## 📡 Buscando os Dados Diretamente do GitHub

Os dados que vamos analisar estão hospedados no **GitHub da pesquisa** — o mesmo repositório onde os pesquisadores da UFCSPA publicaram o código e os dados do artigo científico.

Isso significa que você não precisa baixar nenhum arquivo. Basta passar o endereço (URL) para o pandas, e ele busca os dados direto da fonte — como um DOI de artigo, mas para dados.

Os dados correspondem a **22 jogadores amadores de futebol**, com biomarcadores coletados logo após uma partida.

```python
# =====================================================
# PASSO 2 — Buscando os dados do GitHub da pesquisa
# =====================================================
# A URL aponta para o arquivo CSV bruto no repositório público.
# O pandas consegue ler CSVs direto de uma URL, sem precisar baixar o arquivo!

url = "https://raw.githubusercontent.com/FBRosito/jornada-infobio-2026/main/dados_atletas_minicurso.csv"

# pd.read_csv() lê o arquivo e transforma em uma tabela (DataFrame)
df = pd.read_csv(url)

# df.head() exibe as 5 primeiras linhas — nosso primeiro olhar nos dados!
print(f'✅ Dados carregados: {df.shape[0]} atletas, {df.shape[1]} variáveis\n')
df.head()
```

## 🔍 Inspecionando os Dados

Antes de qualquer análise, um bom cientista de dados **inspeciona o material**. Na bancada molhada, você checaria a pureza de um reagente. Aqui, verificamos:

- **Quantas amostras e variáveis** temos?
- **Que tipo de dado** está em cada coluna (número, texto)?
- **Existem valores ausentes** (campos em branco que poderiam distorcer a análise)?
- **Qual a magnitude** de cada biomarcador — mínimo, máximo, média?

Duas funções cobrem tudo isso em segundos.

```python
# =====================================================
# PASSO 3 — Inspecionando a qualidade e estrutura dos dados
# =====================================================

# df.info() mostra o 'laudo técnico' da tabela:
# tipo de cada coluna e se há valores ausentes (non-null count)
print('--- ESTRUTURA DA TABELA (df.info()) ---')
df.info()

print('\n--- ESTATÍSTICAS DOS BIOMARCADORES (df.describe()) ---')
# df.describe() calcula automaticamente: contagem, média, desvio padrão,
# mínimo, quartis e máximo — tudo de uma vez!
df.describe().round(2)
```

## 📊 Seu Primeiro Gráfico Científico

Chegou o momento mais esperado: **transformar números em imagem**.

Vamos cruzar dois biomarcadores que contam histórias opostas sobre o atleta:

- **CK (Creatina Quinase)** → quanto maior, mais dano muscular mecânico
- **Cortisol** → quanto maior, mais estresse sistêmico e hormonal

Cada ponto no gráfico é um atleta. A posição dele no espaço CK × Cortisol revela em qual estado fisiológico ele se encontrava logo após o jogo.

> 🧠 Observe: existem atletas espalhados pelo gráfico de formas diferentes. No próximo módulo, você vai aprender a identificar padrões nessa nuvem de pontos usando estatística e inteligência artificial.

```python
# =====================================================
# PASSO 4 — Criando o primeiro gráfico de dispersão
# =====================================================
# Um scatter plot posiciona cada atleta como um ponto
# no espaço definido por dois biomarcadores.
# É a forma mais direta de ver relações entre variáveis!

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='CK_UL',          # eixo X: dano muscular mecânico
    y='Cortisol_ugdL',  # eixo Y: estresse sistêmico
    s=100,
    color='steelblue'
)

# Sempre identifique os eixos — um gráfico sem legenda não é científico!
plt.title('Meu Primeiro Gráfico Científico — CK vs Cortisol', fontsize=13, fontweight='bold')
plt.xlabel('CK — Creatina Quinase (U/L)', fontsize=11)
plt.ylabel('Cortisol (µg/dL)', fontsize=11)

plt.tight_layout()
plt.show()

print('🎉 Parabéns! Você acabou de gerar seu primeiro gráfico científico.')
print('   Guarde este notebook — ele marca o início da sua jornada em Informática Biomédica.')
```
