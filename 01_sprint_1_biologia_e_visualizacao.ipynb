```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Carregando os dados dos 22 atletas direto do GitHub
url = "https://raw.githubusercontent.com/FBRosito/jornada-infobio-2026/main/dados_atletas_minicurso.csv"
df = pd.read_csv(url)

print(f'Dataset carregado: {df.shape[0]} atletas, {df.shape[1]} variáveis\n')
df.head()
```

```python
# Resumo estatístico — mínimo, máximo, média de cada biomarcador
df.describe().round(2)
```

## 🔬 O que esses números significam biologicamente?

Antes de plotar, vamos entender o que estamos vendo:

### CK (Creatina Quinase) — eixo X
A CK é uma enzima liberada pelos músculos quando as fibras musculares se rompem durante o exercício intenso. Valores altos indicam **dano mecânico**: o atleta "destruiu" músculo durante o jogo — o que é normal, mas precisa ser monitorado para evitar lesões.

Valores de referência para atletas após treino intenso: **200–1000 U/L**. Acima de 1000 pode indicar rabdomiólise (destruição muscular severa).

### Cortisol — eixo Y
O Cortisol é o "hormônio do estresse". Durante o exercício, o corpo libera Cortisol para mobilizar energia. Mas Cortisol cronicamente elevado **suprime o sistema imune e compromete a recuperação**.

Valores de referência pós-exercício: **8–25 µg/dL**.

### Por que cruzar os dois?
Um atleta com CK baixa **e** Cortisol alto é um sinal de alerta: não há dano muscular visível, mas o estresse sistêmico está elevado. Isso é o que chamamos de **"Risco Silencioso"** — passaria despercebido se analisássemos cada marcador separadamente.

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x='CK_UL',
    y='Cortisol_ugdL',
    hue='Posicao',
    s=120,
    palette='Set2'
)

plt.title('CK vs Cortisol — Resposta Pós-Jogo dos Atletas', fontsize=14, fontweight='bold')
plt.xlabel('CK — Creatina Quinase (U/L)\n[Dano Muscular Mecânico]', fontsize=11)
plt.ylabel('Cortisol (µg/dL)\n[Estresse Sistêmico]', fontsize=11)
plt.legend(title='Posição', fontsize=10)
plt.tight_layout()
plt.show()
```

### 💡 Guia de Consulta Rápida (Cheat Sheet)

* **CK (Creatina Quinase):** Representa o **dano muscular mecânico**.
* **Cortisol:** Representa o **estresse metabólico e sistêmico**.
* **LDH (Lactato Desidrogenase):** Representa o nível de **destruição e desgaste celular geral**.
* **PCR (Proteína C-Reativa):** Representa o nível de **inflamação sistêmica**.
* **Testosterona:** Representa o potencial de **anabolismo e recuperação**.
